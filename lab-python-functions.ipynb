```python
products = ["t-shirt", "mug", "hat", "book", "keychain"]

def initialize_inventory(products):
    inventory = {}
    for i in products:
        cantidad = int(input(f"Ingresa la cantidad disponible para '{i}': "))
        inventory[i] = cantidad
    return inventory

inventory = initialize_inventory(products)
```

```python
def get_customer_orders():
    customer_orders = set() 
    seguir_comprando = input("¿Quieres comprar algo ? yes or no: ").lower()
    while seguir_comprando == "yes":
        product = input("Introduce lo que vas a comprar: ")
        customer_orders.add(product)
        seguir_comprando = input("¿Quieres seguir comprando? yes or no: ").lower()
    
    return customer_orders


```

```python
def update_inventory(customer_orders , inventory):
    for i in customer_orders:
        inventory[i] -= 1
```

```python
def calculate_order_statistics(customer_orders, products):
    order_statistics = len(customer_orders)
    percentage_of_unique = (len(customer_orders) / len(products)) * 100
    return order_statistics, percentage_of_unique
```

```python
def print_order_statistics(order_statistics):
    total_products_ordered, percentage_unique_products = order_statistics
    print(f"Total de productos pedidos: {total_products_ordered}")
    print(f"Porcentaje de productos únicos pedidos: {percentage_unique_products}%")
    
```

```python
def print_updated_inventory(inventory):
    for product, cantidad in inventory.items():
        print(f"{product}: {cantidad}")
```

```python
products = ["t-shirt", "mug", "hat", "book", "keychain"]

inventory = initialize_inventory(products)

customer_orders = get_customer_orders()

update_inventory(customer_orders, inventory)

order_statistics = calculate_order_statistics(customer_orders, products)

print_order_statistics(order_statistics)

print_updated_inventory(inventory)
```
